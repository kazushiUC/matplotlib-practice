```python
fig, ax = plt.subplots(
    2,
    3,
    facecolor="white",
    figsize=(18, 10)
)
```

```python
years = [2021, 2022, 2023, 2024, 2025]
axes = ax.ravel()

for i, year in enumerate(years):
    df_year = df_imports[df_imports["year"] == year]
    df_top10 = df_year.nlargest(10, "import_value")

    axes[i].barh(
        df_top10["Commodity"],
        df_top10["import_million"]
    )

    axes[i].set_xlabel("Commodity")
    axes[i].set_ylabel("import value (million USD)")
    axes[i].set_title(f"Top 10 Imported Commodities in Timor-Leste, {year}")
    axes[i].grid()
```

```python
import matplotlib.pyplot as plt

years = [2021, 2022, 2023, 2024, 2025]

# Create a 2 × 3 subplot grid
fig, ax = plt.subplots(
    2, 3,
    figsize=(18, 10),
    facecolor="white"
)

axes = ax.ravel()

for i, year in enumerate(years):

    # Filter data by year
    df_year = df_imports[df_imports["year"] == year]

    # Select top 10 commodities
    df_top10 = df_year.nlargest(10, "import_value")

    # Create horizontal bar chart
    axes[i].barh(
        df_top10["Commodity"],
        df_top10["import_million"]
    )

    # Improve readability
    axes[i].invert_yaxis()
    axes[i].set_xlabel("Import Value (Million USD)")
    axes[i].set_title(f"Top 10 Imports ({year})")
    axes[i].grid(axis="x", alpha=0.3)

# Hide unused subplot
axes[5].set_visible(False)

plt.tight_layout()
plt.show()
```
